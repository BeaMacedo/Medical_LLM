# **Mixed Dataset for Single-Phase Fine-Tuning**

The three datasets used for both training and validation (MedQA, Medicines, and DDIs) were mixed to the single-phase FT, resulting in two datasets (Train and Dev).

```python
import json
import random
from pathlib import Path
```

## **Train**

```python
DATASET_PATHS = [
    Path(r"../MedicalQA/Data_Generation/MedQA/Train/MedQA_train_final_UPDATED.jsonl"),
    Path(r"../Drugs/Data_Generation/DDIs/data_finetuning/final_datasets/final_TRAIN.jsonl"),
    Path(r"../Drugs/Data_Generation/Medicines/data_finetuning/final_datasets/clean/Finetuning_medicines_TRAIN.jsonl"),
]

OUTPUT_PATH = Path("mixed_TRAIN.jsonl")

all_records = []

for path in DATASET_PATHS:
    records = []
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line:
                records.append(json.loads(line))
    print(f"{path.name}: {len(records)} examples")
    all_records.extend(records)

print(f"Total before shuffling: {len(all_records)} examples")

SEED = 42
random.seed(SEED)
random.shuffle(all_records)

print(f"Dataset shuffled with seed={SEED}")
print(f"Total examples: {len(all_records)}")

with open(OUTPUT_PATH, 'w', encoding='utf-8') as f:
    for record in all_records:
        f.write(json.dumps(record, ensure_ascii=False) + '\n') #writes back in jsonl format #ensure_ascii=False preserves special characters like accents and medical symbols (µg, °C, etc.) instead of converting them into \uXXXX sequences

#print(f"File saved to: {OUTPUT_PATH.resolve()}") 
print(f"Lines written: {len(all_records)}")
```

## **DEV**

```python
DEV_PATHS = [ 
    Path(r"../MedicalQA/Data_Generation/MedQA/Train/MedQA_dev_UPDATED.jsonl"),
    Path(r"../Drugs/Data_Generation/DDIs/data_finetuning/final_datasets/final_DEV.jsonl"),
    Path(r"../Drugs/Data_Generation/Medicines/data_finetuning/final_datasets/clean/Finetuning_medicines_DEV.jsonl"),
]

DEV_OUTPUT_PATH = Path("mixed_DEV.jsonl")

dev_records = []

for path in DEV_PATHS:
    records = []
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line:
                records.append(json.loads(line))
    print(f"{path.name}: {len(records)} examples")
    dev_records.extend(records)

print(f"Total before shuffling: {len(dev_records)} examples")

SEED = 42
random.seed(SEED)
random.shuffle(dev_records)

print(f"Dataset shuffled with seed={SEED}")

with open(DEV_OUTPUT_PATH, 'w', encoding='utf-8') as f:
    for record in dev_records:
        f.write(json.dumps(record, ensure_ascii=False) + '\n')

#print(f"File saved to: {DEV_OUTPUT_PATH.resolve()}")
print(f"Lines written: {len(dev_records)}")
```

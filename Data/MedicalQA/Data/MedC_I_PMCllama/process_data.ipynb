```python
from datasets import load_dataset
import pandas as pd

print("1. Loading dataset 'axiong/pmc_llama_instructions'...")

dataset = load_dataset("axiong/pmc_llama_instructions", split="train")

unique_sources = set(dataset['source'])
print(f"Sources found in dataset: {unique_sources}")
```

```python
import os

output_folder = "MedQA"
output_filename = "medqa_data.json"

if not os.path.exists(output_folder):
    os.makedirs(output_folder)
    print(f" -> Folder '{output_folder}' created successfully.")

print("\n--- Filtering MedQA data... ---")
medqa_data = dataset.filter(lambda example: example['source'] == 'medqa_train')
print(f" -> Found {len(medqa_data)} examples.")

file_path = os.path.join(output_folder, output_filename)

medqa_data.to_json(file_path)

print(f"\nDone, data saved to: {file_path}")
```
